# SmartHire — EDA
Uses the actual Resume.csv dataset supplied for the project.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
resume_df = pd.read_csv(
    "../data/raw/Resume.csv"
)

print("Resume Dataset Shape:", resume_df.shape)

resume_df.head()
```

```python
print("Resume Dataset Columns:")
print(resume_df.columns.tolist())
```

```python
print("Dataset Information:")
resume_df.info()
```

```python
print("Missing Values:")
print(resume_df.isnull().sum())
```

```python
print("Duplicate Rows:", resume_df.duplicated().sum())
print(
    "Duplicate Resume Text:",
    resume_df["Resume_str"].duplicated().sum()
)
```

```python
print(
    "Number of Resume Categories:",
    resume_df["Category"].nunique()
)
```

```python
category_counts = resume_df["Category"].value_counts()

print(category_counts)
```

```python
plt.figure(figsize=(14, 7))

category_counts.plot(
    kind="bar"
)

plt.title("Resume Category Distribution")
plt.xlabel("Resume Category")
plt.ylabel("Number of Resumes")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

```python
jobs_df = pd.read_csv(
    "../data/processed/jobs_clean.csv"
)

print(
    "Job Dataset Shape:",
    jobs_df.shape
)

jobs_df.head()
```

```python
print("Job Dataset Columns:")
print(jobs_df.columns.tolist())
```

```python
print("Job Dataset Information:")

jobs_df.info()
```

```python
print("Job Dataset Missing Values:")

print(
    jobs_df.isnull().sum()
)
```

```python
print(
    "Duplicate Job Rows:",
    jobs_df.duplicated().sum()
)
```

```python
print("Top 20 Job Titles:")

print(
    jobs_df["jobtitle"]
    .value_counts()
    .head(20)
)
```

```python
plt.figure(figsize=(14, 7))

jobs_df["jobtitle"].value_counts().head(20).plot(
    kind="bar"
)

plt.title("Top 20 Job Titles")
plt.xlabel("Job Title")
plt.ylabel("Number of Jobs")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

```python
print("Top 20 Companies:")

print(
    jobs_df["company"]
    .value_counts()
    .head(20)
)
```

```python
print("Top 20 Job Locations:")

print(
    jobs_df["joblocation_address"]
    .value_counts()
    .head(20)
)
```

```python
print("=" * 60)
print("SMART HIRE - EDA SUMMARY")
print("=" * 60)

print(
    "Resume Dataset:",
    resume_df.shape
)

print(
    "Resume Categories:",
    resume_df["Category"].nunique()
)

print(
    "Job Dataset:",
    jobs_df.shape
)

print(
    "Unique Job Titles:",
    jobs_df["jobtitle"].nunique()
)

print(
    "Unique Companies:",
    jobs_df["company"].nunique()
)

print("=" * 60)
```
